```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Sample agriculture sensor data
data = {
    "Date": [
        "2026-10-20",
        "2026-10-21",
        "2026-10-22",
        "2026-10-23",
        "2026-10-24",
        "2026-10-25"
    ],
    "Temperature": [28, 30, 27, 32, 29, 26],
    "Humidity": [65, 60, 70, 55, 68, 75],
    "Soil_Moisture": [40, 35, 55, 30, 45, 60],
    "Crop_Growth": [12, 13, 14, 15, 16, 17]
}

# Organize data using Pandas
df = pd.DataFrame(data)

# Display the sensor data
print("SMART AGRICULTURE SENSOR DATA")
display(df)
```

```python
# Convert sensor columns into NumPy arrays
temperature = np.array(df["Temperature"])
humidity = np.array(df["Humidity"])
soil_moisture = np.array(df["Soil_Moisture"])

print("NUMPY SENSOR ANALYSIS")

print("Average Temperature:",
      np.mean(temperature), "°C")

print("Maximum Temperature:",
      np.max(temperature), "°C")

print("Minimum Temperature:",
      np.min(temperature), "°C")

print("Average Humidity:",
      np.mean(humidity), "%")

print("Average Soil Moisture:",
      np.mean(soil_moisture), "%")
```

```python
print("SENSOR DATA SUMMARY")

# Display basic statistics
display(df.describe())

# Find readings where soil moisture is below 40%
low_moisture = df[df["Soil_Moisture"] < 40]

print("Readings with soil moisture below 40%:")
display(low_moisture)
```

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Select sensor measurements for analysis
X = df[
    ["Temperature", "Humidity", "Soil_Moisture"]
]

# Standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Apply K-Means clustering
model = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

df["Cluster"] = model.fit_predict(X_scaled)

print("CROP CONDITION GROUPING")
display(df)
```

```python
# Plot temperature and humidity
plt.figure(figsize=(10, 5))

plt.plot(
    df["Date"],
    df["Temperature"],
    marker="o",
    label="Temperature (°C)"
)

plt.plot(
    df["Date"],
    df["Humidity"],
    marker="s",
    label="Humidity (%)"
)

plt.title("Agriculture Sensor Monitoring")
plt.xlabel("Date")
plt.ylabel("Reading")
plt.legend()
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Select agriculture sensor measurements
heatmap_data = df[
    ["Temperature", "Humidity", "Soil_Moisture", "Crop_Growth"]
].copy()

# Rename columns for display
heatmap_data.columns = [
    "Temperature (°C)",
    "Humidity (%)",
    "Soil Moisture (%)",
    "Crop Growth (cm)"
]

# Normalize values so different units can be compared
normalized_data = (
    heatmap_data - heatmap_data.min()
) / (
    heatmap_data.max() - heatmap_data.min()
)

# Create colourful heatmap
plt.figure(figsize=(11, 6))

sns.heatmap(
    normalized_data,
    annot=heatmap_data.round(1),
    fmt=".1f",
    cmap="RdYlGn",
    linewidths=1,
    linecolor="white",
    cbar_kws={"label": "Normalized Value (0–1)"}
)

plt.title(
    "Smart Agriculture Data Monitoring Heatmap",
    fontsize=16,
    fontweight="bold",
    pad=20
)

plt.xlabel("Agriculture Parameters", fontsize=12)
plt.ylabel("Sensor Reading", fontsize=12)

plt.tight_layout()
plt.show()
```
